**Rules for naming variables in Python**

1. A name must start with a letter (a to z, A to Z) or an underscore (`_`).
2. A name cannot start with a digit.
3. After the first character, a name may contain only letters, digits and underscores.
4. A name cannot contain spaces or symbols such as `-`, `(`, `)`, `&`, `@` or `$`.
5. A name cannot be a Python keyword such as `if`, `for`, `True` or `class`.
6. Names are case sensitive, so `Sale_price_2021` and `sale_price_2021` are two different variables.

**Acceptable variable names**

| Variable name | Reason it is acceptable |
|---|---|
| `grade_test` | Starts with a letter and uses only letters and an underscore. |
| `covid_19_cases` | Starts with a letter. Underscores join the words, and digits after the first character are allowed. |
| `term2` | Starts with a letter. The digit comes after the first character. |
| `Sale_price_2021` | Starts with a capital letter, which is allowed. Contains only letters, digits and underscores. |
| `miva_learners` | Starts with a letter. An underscore joins the two words instead of a space. |

**Unacceptable variable names**

| Variable name | Reason it is not acceptable |
|---|---|
| `test(grade)` | Brackets are not allowed. Python reads `test(grade)` as a call to a function named `test`. |
| `covid-19-cases` | The hyphen is the minus operator, so Python reads this as `covid` minus `19` minus `cases`. |
| `2term` | A variable name cannot start with a digit. |
| `2021sales_price` | A variable name cannot start with a digit. |
| `miva&learners` | `&` is an operator in Python and is not allowed inside a name. |

```python
# Demonstration: valid names work
grade_test = 75
covid_19_cases = 120
term2 = "Second term"
Sale_price_2021 = 2500.50
miva_learners = 300
print(grade_test, covid_19_cases, term2, Sale_price_2021, miva_learners)

# Demonstration: invalid names raise a SyntaxError
for bad in ["test(grade) = 1", "covid-19-cases = 1", "2term = 1",
            "2021sales_price = 1", "miva&learners = 1"]:
    try:
        exec(bad)
    except SyntaxError as e:
        print(f"{bad:<22} -> SyntaxError: {e.msg}")
```
