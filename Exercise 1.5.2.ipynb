The number of confirmed COVID-19 cases in selected countries is shown in the table below:

|             |                                |

| Burundi     | 694                            |
| Ethiopia    | 113295                         |
| Ghana       | 52198                          |

In this task, we are going to practice using and manipulating dictionaries.

1. Create a dictionary for the given data and name it `COVID_19`

2. How many keys are in the data? assign your answer to a variable called `num_keys`.

3. Remove Nigeria from the data

4. Add Tanzania with the value $509$

#### Question 1: Create a dictionary for the given data and name it `COVID_19`

```python
### CG T1_1
### Write your code below:

# Each country is a key and its number of confirmed cases is the value
COVID_19 = {
    "Burundi": 694,
    "Ethiopia": 113295,
    "Ghana": 52198,
    "Kenya": 88380,
    "Nigeria": 68937,
    "Rwanda": 6129,
    "Somalia": 4525,
    "South Sudan": 3166,
    "Sudan": 19196,
    "Uganda": 22499,
}
print(COVID_19)
```

#### Question 2: How many keys are in the data? assign your answer to a variable called `num_keys`.

```python
### CG T1_2
### Write your code below:

num_keys = len(COVID_19)   # len() counts the keys in a dictionary
print("Number of keys:", num_keys)
```

#### Question 3: Remove Nigeria from the dataset

```python
### CG T1_3
### Write your code below:

del COVID_19["Nigeria"]   # removes the key "Nigeria" and its value
print(COVID_19)
```

#### Question 4: Add Tanzania to the dataset. Tanzania recorded $509$ cases of COVID-19.

```python
### CG T1_4
### Write your code below:

COVID_19["Tanzania"] = 509   # assigning to a new key adds it to the dictionary
print(COVID_19)
```

## Task 2

Consider the dictionary:

```python
### CG T2_init
### DON'T DELETE THIS CODE!

food = {
    "fruits": ["apples", "oranges", "pears", "mangoes"],
    "vegetables": ["tomatoes", "lettuce", "spinach", "green peppers"],
    "meat": ["chicken", "fish", "beef", "ostrich"],
    "dairy": ["yogurt", "milk", "cheese", "ice-cream"]
}
```

#### Question 1: How many keys does the dictionary `food` have?

```python
### CG T2_1
### Write your code below:

print("Number of keys in food:", len(food))
print("The keys are:", list(food.keys()))
```

#### Question 2: List the values in `food`

```python
### CG T2_2
### Write your code below:

print(list(food.values()))
```

#### Question 3: How do you access `"spinach"` from `food`?

```python
### CG T2_3
### Write your code below:

# "spinach" is at index 2 of the list stored under the key "vegetables"
print(food["vegetables"][2])
```

#### Question 4: How do you add `"grapes"` to the list of fruits in `food`?

```python
### CG T2_4
### Write your code below:

food["fruits"].append("grapes")   # append() adds an item to the end of a list
print(food["fruits"])
```
